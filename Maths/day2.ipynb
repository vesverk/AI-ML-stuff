```python
#common probability distribution
    #gaussian distribution

import numpy as np
import matplotlib.pyplot as plt

mu,sigma  = 0,1
x = np.linspace(-4,4,100)
y = (1/(np.sqrt(2*np.pi * sigma**2))) * np.exp(-0.5*((x-mu)/sigma)**2) #gaussian formula 

plt.plot(x,y)
plt.show()
```

```python
#bernoli distribution

from scipy.stats import binom

n,p = 10,0.5
x = np.arange(0,n+1)
y = binom.pmf(x,n,p)
plt.bar(x,y, color="green")
plt.show()
```

```python
#poisson distribtuin : models the numver of event in a fixed interval of time or space

from scipy.stats import poisson

lam = 3
x = np.arange(0,10)
y = poisson.pmf(x,lam)
plt.bar(x,y,color="pink")
plt.show()
```

```python
def bayes_theorem(prior , sensitivity , specificity):
    evidance = (sensitivity*prior) + ((1-specificity)*(1-prior))
    posterior = (sensitivity*prior)/evidance
    return posterior

result = bayes_theorem(prior=0.01 , sensitivity=0.95 , specificity=0.90)
print(result)
```
