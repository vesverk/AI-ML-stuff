```python
#gradiant descent is algo used to optimize a  fucntion 
#update parameter in the derection of the negative gradiant to find th eminimum
# update rule  --> theta = theta-alpha*nabla*f(theta)

#theta = parameter of model
#alpha = learning rate(step size)


def gradient_decent(X,y,theta,learning_rate,iteration):
    m = len(y)
    for _ in range(iteration):
        predictions = np.dot(X, theta)
        errors = predictions -y
        gradients = (1/m) * np.dot(X.T, errors)
        theta -= learning_rate * gradients
    return theta

x= np.array([[1,1],[1,2],[1,3]])
y= np.array([2,2.5,3.5])
theta = np.array([0.1,0.1])
learning_rate = 0.1
iteration = 1000

optimized_theta = gradient_decent(x,y,theta,learning_rate,iteration)

print(optimized_theta)
```

```python
#quadratic  = f(x) = x^2

import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2

def gradient(x):
    return 2*x

x = 4
learning_rate = 0.1
iteration = 20

x_history = [x]

for i in range(iteration):
    x = x-learning_rate*gradient(x)
    x_history.append(x)


x_values = np.linspace(-5,5,200)
y_values = f(x_values)

plt.plot(x_values,y_values)

x_points = np.array(x_history)
y_points = f(x_points)

plt.scatter(x_points,y_points)

plt.plot(x_points,y_points)
plt.xlabel("x")
plt.ylabel("f(x) = x²")
plt.title("Gradient Descent on Quadratic Function")

plt.grid()
plt.show()





```
