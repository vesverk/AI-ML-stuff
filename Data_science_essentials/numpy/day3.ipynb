```python
# boradcasting


#pytho loops are slow 

#so to fix it we can use vectorization like we can diectly sat to arr that change ot power of two

import numpy as np
import time
start = time.time()

arr = np.array([1, 2, 3, 4, 5])
result = []

# Using a loop to square each element (slow)
for num in arr:
    result.append(num ** 2)

print(result)  # Output: [1, 4, 9, 16, 25]
end = time.time()

print(end - start)
```

```python
#vectorized
import numpy as np
import time

start = time.time()
arr = np.array([1, 2, 3, 4, 5])
result = arr ** 2  # Vectorized operation
print(result)  # Output: [1 4 9 16 25]
end = time.time()

print(end-start)
```

```python
result + 10  #this is broadcasting : brodacating 10 is added to all elements without using extra memory
```
